```python
# Import libraries
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.load_data import load_train_data, load_test_data
from src.clean_data import get_data_quality_report

# Set display options
pd.set_option('display.max_columns', None)
plt.style.use('seaborn-v0_8-darkgrid')
%matplotlib inline
```

## 1. Load the Data

**Note:** Before running this notebook, download the Kaggle House Prices dataset from:
https://www.kaggle.com/c/house-prices-advanced-regression-techniques/data

Place `train.csv` and `test.csv` in the `data/raw/` directory.

```python
# Load training data
try:
    train_df = load_train_data()
    print(f"Training data shape: {train_df.shape}")
except FileNotFoundError as e:
    print(f"Error: {e}")
    print("Please download the dataset and place it in data/raw/")
```

## 2. Explore Dataset Structure

```python
# Display first few rows
train_df.head()
```

```python
# Display basic information
train_df.info()
```

```python
# Display statistical summary
train_df.describe()
```

## 3. Analyze Missing Values

```python
# Get data quality report
quality_report = get_data_quality_report(train_df)
print("\nColumns with missing values:")
quality_report[quality_report['Missing_Count'] > 0]
```

```python
# Visualize missing values
missing_data = train_df.isnull().sum()
missing_data = missing_data[missing_data > 0].sort_values(ascending=False)

plt.figure(figsize=(12, 6))
missing_data.plot(kind='bar')
plt.title('Missing Values by Column')
plt.ylabel('Count of Missing Values')
plt.xlabel('Column')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## 4. Visualize Key Features

```python
# Distribution of target variable (SalePrice)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
axes[0].hist(train_df['SalePrice'], bins=50, edgecolor='black')
axes[0].set_title('Distribution of Sale Price')
axes[0].set_xlabel('Sale Price')
axes[0].set_ylabel('Frequency')

# Log-transformed histogram
axes[1].hist(np.log1p(train_df['SalePrice']), bins=50, edgecolor='black', color='orange')
axes[1].set_title('Distribution of Log(Sale Price)')
axes[1].set_xlabel('Log(Sale Price)')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

print(f"SalePrice Statistics:")
print(train_df['SalePrice'].describe())
```

```python
# Relationship between key features and SalePrice
features_to_plot = ['GrLivArea', 'TotalBsmtSF', 'OverallQual', 'YearBuilt']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.ravel()

for idx, feature in enumerate(features_to_plot):
    axes[idx].scatter(train_df[feature], train_df['SalePrice'], alpha=0.5)
    axes[idx].set_xlabel(feature)
    axes[idx].set_ylabel('SalePrice')
    axes[idx].set_title(f'SalePrice vs {feature}')

plt.tight_layout()
plt.show()
```

## 5. Examine Correlations

```python
# Correlation with SalePrice
numerical_features = train_df.select_dtypes(include=[np.number]).columns
correlations = train_df[numerical_features].corr()['SalePrice'].sort_values(ascending=False)

print("Top 10 Features Correlated with SalePrice:")
print(correlations.head(11))  # 11 to include SalePrice itself

print("\nBottom 10 Features Correlated with SalePrice:")
print(correlations.tail(10))
```

```python
# Correlation heatmap for top features
top_features = correlations.head(11).index
plt.figure(figsize=(12, 10))
sns.heatmap(train_df[top_features].corr(), annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Heatmap of Top Features')
plt.tight_layout()
plt.show()
```
